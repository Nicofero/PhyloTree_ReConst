# **Benchmarking metrics**

After obtaining the RF distances between every reconstructed tree and its ground truth, we can now represent the obtained results to see how the method works.

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import pandas as pd
import numpy as np
from collections import defaultdict
```

# **RF-Distance**

We have 3 distance datasets for every specie, so to save them we will use a dictionary.

```python
# Initialization
bos_taurus = {}
homo_sapiens = {}
penicillium_digitatum = {}
puccinia_graminis = {}
schizosaccharomyces_pombe = {}
all_species = [bos_taurus, homo_sapiens, penicillium_digitatum, puccinia_graminis, schizosaccharomyces_pombe]
methods = ['qa', 'qa_adv2', 'hy', 'qaoa', 'qrao', 'pce']
species_names = ['Bos_taurus','Homo_sapiens','Puccinia_graminis','Penicillium_digitatum','Schizosaccharomyces_pombe']

for i,specie in enumerate(all_species):
    for method in methods:
        specie[method] = pd.read_csv(f'./{species_names[i]}/nmcut{method}_RF-distance.csv')

bos_taurus['qa'].head()
```

We have the same conclusions that we had before, for the quantum methods it exists a clear downward tendency as the number of sequences grows. This is true for both advantage versions. However this downward tendency doesn't appears in the hybrid approach meaning it has to do with the nature of the quantum methods. It probably has something to do with the way the embeddings with in the Advantage QPU.

The branch size, as i suspected before, doesn't make a difference for the precision. It seems that the precision is independent from the branch length, however it seems to begin to grow when it is large. Though I don't have enough data to confirm this theory, specially for the QPU methods.

# **Time**

I have time data for the methods, however, for the hybrid version, the time is irrelevant as it operates using a max time of computation and then yields the best result after that time, thus meaning the time it takes to create one tree is $\max_{time}\cdot N_{probs}$. It does make sense to compare for the QPU methods, so we can do that.

```python
# Initialization
bos_taurus_time = {}
homo_sapiens_time = {}
penicillium_digitatum_time = {}
puccinia_graminis_time = {}
schizosaccharomyces_pombe_time = {}
all_species_time = [bos_taurus_time, homo_sapiens_time, penicillium_digitatum_time, puccinia_graminis_time, schizosaccharomyces_pombe_time]
methods = ['', '_adv2']
species_names = ['Bos_taurus','Homo_sapiens','Puccinia_graminis','Penicillium_digitatum','Schizosaccharomyces_pombe']

for i,specie in enumerate(all_species_time):
    for method in methods:
        specie[method] = pd.read_csv(f'./{species_names[i]}/timer{method}_{species_names[i]}.csv')

bos_taurus_time[''].head()
```

```python
# Create 2x3 grid
fig, axes = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(20, 10),
    sharex=True,
    sharey=True
)

axes = axes.flatten()

# Plot only first 5 axes
for ax, species, name in zip(axes[:5], all_species_time, species_names):

    for i, method in enumerate(methods):
        ax.scatter(
            species[method]['size'],
            species[method]['time']/1000,
            edgecolor='k',
            s=100,
            c=colors[i],
            marker=shape[i],
            label= method if method != '' else 'adv1'
        )

    ax.set_title(name, fontsize=18)
    ax.grid(color='gray', linestyle='--', linewidth=0.4)

# Hide unused subplot (bottom-right)
axes[5].axis('off')

# Global labels
fig.supxlabel('Number of sequences', fontsize=24)
fig.supylabel('Time (s)', fontsize=24)

# One global legend
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc='upper center',
    ncol=len(methods),
    fontsize=20
)

plt.tight_layout(rect=[0, 0, 1, 0.9])
plt.show()
```
